**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
import pandas as pd
df = pd.read_parquet("../data/laps_2023_2025.parquet") 
print(df.shape)
print(df["session_key"].nunique(), "races")
print(df.groupby("year")["session_key"].nunique())
print(df["driver_number"].nunique(), "drivers")
df.head()
```

```python
rc = pd.read_parquet("../data/race_control_2023_2025.parquet")
print(df["name_acronym"].nunique(), "drivers")
print(rc.shape[0], "race control rows")
print(df.groupby("year")["meeting_name"].unique().to_string())
```

**Answer:** The dataset has **75,181 rows and 20 columns**. One row represents **one lap by one driver in one Formula 1 Grand Prix race**. Sprint races, practice and qualifying are not included. It covers the **2023-2025 seasons: 68 races** (22 in 2023, 23 in 2024 and 23 in 2025) and 28 different drivers.

The data comes from the OpenF1 API and was assembled by joining four of its tables: lap timing (lap time, sector times, pit out-lap flag), stints (tyre compound and tyre age at the start of each stint), drivers (driver code and team) and meetings/sessions (race, circuit and date). The tyre age of every lap was computed from the stint information. A second, smaller table with 6286 rows contains the race control messages of the same races (flags, safety car and incidents), which will be used to identify neutralised laps.

The seasons were limited to 2023-2025 because OpenF1 provides free historical data only from 2023, and because this period falls within a single technical regulation era (the cars and tyres introduced in 2022), which keeps the tyre degradation behaviour comparable. **c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
#2. Limited Y-axis

#The graph is clearer, but lap times from different circuits are not directly comparable.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="tyre_age",
    y="lap_duration",
    hue="compound",
    alpha=0.3
)

plt.title("Lap time vs tyre age by compound")
plt.xlabel("Tyre age (laps)")
plt.ylabel("Lap duration (seconds)")
plt.ylim(60, 250)
plt.show()
```

Most of the lap times are at the bottom of the graph because some lap times are extremely long. We limited the Y-axis to make it easier to read, but comparing different circuits is still not very useful.

```python
# 3. Monza 2024 – All drivers

#Circuit differences are removed, but driver and car performance still affect lap times.
race_df = df[
    (df["year"] == 2024) &
    (df["circuit"] == "Monza") &
    (df["lap_duration"].between(60, 180))
]

plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=race_df,
    x="tyre_age",
    y="lap_duration",
    hue="compound",
    alpha=0.5
)

plt.title("Lap time vs tyre age – Monza 2024")
plt.xlabel("Tyre age (laps)")
plt.ylabel("Lap duration (seconds)")
plt.grid(alpha=0.2)
plt.show()
```

This graph is easier to understand because we only look at one race. However, different drivers and cars still affect the lap times, so it is hard to see how much tyre age actually matters.

```python
#4. Monza 2024 – One driver

#Focusing on one driver makes comparisons more meaningful, although fuel load and race conditions still influence lap times.
import matplotlib.pyplot as plt
import seaborn as sns

# Exclude unusually slow laps
driver_clean = driver_df[
    driver_df["lap_duration"].between(80, 95)
].copy()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=driver_clean,
    x="tyre_age",
    y="lap_duration",
    hue="stint_number",
    style="compound",
    markers=True,
    dashes=False,
    estimator=None
)

plt.title("Max Verstappen – Lap Time by Tyre Age (Monza 2024)")
plt.xlabel("Tyre Age (laps)")
plt.ylabel("Lap Time (seconds)")
plt.grid(alpha=0.3)
plt.show()
```

Verstappen's lap times stay quite similar during each stint. The later stints seem faster, but this could also be because the car gets lighter as fuel is used.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select Monza 2024
compound_df = df[
    (df["year"] == 2024) &
    (df["circuit"] == "Monza") &
    (df["compound"].isin(["MEDIUM", "HARD"])) &
    (df["lap_duration"].between(80, 95)) &
    (df["is_pit_out_lap"] != True)
].copy()

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=compound_df,
    x="compound",
    y="lap_duration",
    order=["MEDIUM", "HARD"]
)

plt.title("Lap Time Distribution by Tyre Compound – Monza 2024")
plt.xlabel("Tyre Compound")
plt.ylabel("Lap Time (seconds)")
plt.grid(axis="y", alpha=0.2)

plt.show()
```

Hard tyres have slightly faster lap times on average, but this does not mean they are actually faster than medium tyres. Other things like fuel load and race strategy also affect the results. We left out soft tyres because there was only one valid lap.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
## TODO

# 1. Missing values
print("=== MISSING VALUES ===")
missing = pd.DataFrame({
    "Missing": df.isna().sum(),
    "Percentage": (df.isna().mean() * 100).round(2)
})
print(missing[missing["Missing"] > 0])

# 2. Duplicate laps
print("\n=== DUPLICATES ===")
duplicates = df.duplicated(
    subset=["session_key", "driver_number", "lap_number"]
).sum()
print("Duplicate laps:", duplicates)

# 3. Invalid or unusual lap times
print("\n=== LAP TIMES ===")
print(df["lap_duration"].describe())

print("Missing lap times:", df["lap_duration"].isna().sum())
print("Non-positive lap times:", (df["lap_duration"] <= 0).sum())
print("Laps over 180 seconds:", (df["lap_duration"] > 180).sum())

# 4. Tyre age
print("\n=== TYRE AGE ===")
print("Missing tyre ages:", df["tyre_age"].isna().sum())
print("Negative tyre ages:", (df["tyre_age"] < 0).sum())

# 5. Tyre compounds
print("\n=== TYRE COMPOUNDS ===")
print(df["compound"].value_counts(dropna=False))

# 6. Race coverage
print("\n=== RACE COVERAGE ===")
print(df.groupby("year")["session_key"].nunique())
```

**Answer:** Overall, our data looks quite good. There are only a few missing values (less than 1% in each column) and no duplicate laps. However, some lap times are extremely long, with the longest being over 2600 seconds. These could be caused by red flags, pit stops or other race situations.

For our analysis, we plan to remove rows where important values are missing and filter out unusually long lap times when needed. We also need to be careful because not every slow lap means there is something wrong with the data.
